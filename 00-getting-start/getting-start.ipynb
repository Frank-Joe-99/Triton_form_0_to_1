# Installation by uv

```shell
uv venv triton
# 进入当前环境
source ./triton/bin/activate

uv pip install triton
```

```shell
uv pip install torch torchvision
uv pip install matplotlib ipykernel
```

```shell
#退出当前环境
deactivate
```
